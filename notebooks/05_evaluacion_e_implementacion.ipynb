# Fase 5. Evaluación

En esta fase se evalúa el modelo teniendo en cuenta el cumplimiento de los criterios de éxito del
problema, se revisa el proceso completo y se decide si se puede pasar a la implementación.

## 5.1 Evaluación de los resultados

Evaluación del modelo en relación con los objetivos y criterios de éxito del negocio.

## 5.2 Revisión del proceso

Revisión de todo el proceso de minería de datos para identificar elementos que puedan mejorarse.

## 5.3 Determinación de los próximos pasos

Decisión sobre si se pasa a la implementación o si se repite alguna fase anterior.

# Fase 6. Implementación

En esta fase el conocimiento obtenido se transforma en acciones dentro del proceso de negocio y se
documentan y presentan los resultados.

## 6.1 Plan de implementación

Estrategia para aplicar los resultados del modelo en la organización.

## 6.2 Monitoreo y mantenimiento

Estrategia para dar seguimiento al modelo y mantenerlo actualizado.

## 6.3 Informe final

Conclusión del proyecto y presentación de los resultados obtenidos.

## 6.4 Revisión del proyecto

Evaluación de lo que se hizo bien y de lo que se debe mejorar.
